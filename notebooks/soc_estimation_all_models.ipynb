```python
# ============================================================================
# 1. Imports and reproducibility
# ============================================================================
import os
import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

import tensorflow as tf
from tensorflow.keras import Model
from tensorflow.keras.layers import (
    Input, Dense, Dropout, LayerNormalization, MultiHeadAttention, Add, Lambda
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style='whitegrid')
plt.rcParams['figure.dpi'] = 120

print('TensorFlow:', tf.__version__)
print('GPUs:', tf.config.list_physical_devices('GPU'))
```

```python
# ============================================================================
# 2. Dataset configuration
# ============================================================================
# Update DATA_DIR to point to your local copy of the NASA Battery Dataset.
# Expected layout:
#   <DATA_DIR>/metadata.csv
#   <DATA_DIR>/data/05000.csv ...
import glob
from pathlib import Path

DATA_DIR = Path('../data/raw')  # <-- update this if your data is elsewhere

meta_files = list(DATA_DIR.glob('**/metadata.csv'))
assert len(meta_files) > 0, f'No metadata.csv found under {DATA_DIR}.
METADATA_PATH = meta_files[0]
DATASET_DIR   = METADATA_PATH.parent

all_csvs = glob.glob(str(DATASET_DIR / '**' / '*.csv'), recursive=True)
csv_map  = {Path(p).name: Path(p) for p in all_csvs}

TRAIN_BATTERIES = [
    'B0005', 'B0007', 'B0028', 'B0032', 'B0034',
    'B0045', 'B0046', 'B0047', 'B0048', 'B0049', 'B0054'
]
VAL_BATTERIES  = ['B0018']
TEST_BATTERIES = ['B0006']

SEQ_LENGTH   = 30
FEATURE_COLS = ['voltage', 'current', 'temperature', 'dt']
TARGET_COL   = 'SOC'

print('Metadata:', METADATA_PATH)
print('CSV files found:', len(csv_map))
```

```python
# ============================================================================
# 3. Load discharge rows from NASA-style files
# ============================================================================
metadata = pd.read_csv(METADATA_PATH)
metadata.columns = [c.lower().strip() for c in metadata.columns]

required_meta = {'type', 'battery_id', 'filename', 'capacity'}
missing = required_meta.difference(metadata.columns)
assert not missing, f'Missing metadata columns: {missing}'

discharge_meta = metadata[metadata['type'].str.lower().eq('discharge')].copy()
discharge_meta['capacity'] = pd.to_numeric(discharge_meta['capacity'], errors='coerce')
discharge_meta = discharge_meta.dropna(subset=['capacity'])

selected_batteries = TRAIN_BATTERIES + VAL_BATTERIES + TEST_BATTERIES
discharge_meta = discharge_meta[discharge_meta['battery_id'].isin(selected_batteries)].copy()

sort_cols = ['battery_id'] + [c for c in ['uid', 'test_id'] if c in discharge_meta.columns]
discharge_meta = discharge_meta.sort_values(sort_cols)
discharge_meta['cycle'] = discharge_meta.groupby('battery_id').cumcount() + 1

rows = []
for _, meta_row in discharge_meta.iterrows():
    file_name = Path(str(meta_row['filename'])).name
    file_path = csv_map.get(file_name)
    if file_path is None or not file_path.exists():
        continue

    df = pd.read_csv(file_path)
    df.columns = [c.lower().strip() for c in df.columns]
    df = df.rename(columns={
        'voltage_measured': 'voltage',
        'current_measured': 'current',
        'temperature_measured': 'temperature',
        'time': 'time',
    })

    needed = ['voltage', 'current', 'temperature', 'time']
    if not all(col in df.columns for col in needed):
        continue

    df = df[needed].copy()
    for col in needed:
        df[col] = pd.to_numeric(df[col], errors='coerce')
    df = df.dropna(subset=needed)

    df['capacity'] = float(meta_row['capacity'])
    df['battery_name'] = meta_row['battery_id']
    df['cycle'] = int(meta_row['cycle'])
    rows.append(df)

assert len(rows) > 0, 'No discharge files were loaded. Check that metadata filenames match the Kaggle CSV files.'

raw_df = pd.concat(rows, ignore_index=True)
raw_df = raw_df.sort_values(['battery_name', 'cycle', 'time']).reset_index(drop=True)

print(f'Rows loaded: {len(raw_df):,}')
print(f'Batteries loaded: {raw_df["battery_name"].nunique()}')
print(raw_df.groupby('battery_name')['cycle'].nunique().sort_index())
raw_df.head()
```

```python
# ============================================================================
# 4. Derive SoC with Coulomb counting and add dt
# ============================================================================
def compute_soc_coulomb(group):
    group = group.sort_values('time').copy()
    dt = group['time'].diff().fillna(0).clip(lower=0)

    charge_consumed_as = (group['current'].abs() * dt).cumsum()
    q_nominal_as = group['capacity'].iloc[0] * 3600.0

    if q_nominal_as <= 0:
        group['SOC'] = np.nan
        group['dt'] = dt
        return group

    group['SOC'] = (1.0 - charge_consumed_as / q_nominal_as).clip(0.0, 1.0) * 100.0
    group['dt'] = dt
    return group

raw_df = raw_df.groupby(['battery_name', 'cycle'], group_keys=False).apply(compute_soc_coulomb)
raw_df = raw_df.dropna(subset=['SOC']).reset_index(drop=True)

print(f'Rows after SoC calculation: {len(raw_df):,}')
print(f'SoC range: {raw_df.SOC.min():.2f}% to {raw_df.SOC.max():.2f}%')

sample = raw_df[(raw_df['battery_name'] == TEST_BATTERIES[0]) & (raw_df['cycle'] == 1)]
fig, ax = plt.subplots(figsize=(9, 3))
ax.plot(sample['time'], sample['SOC'], color='steelblue')
ax.set_title(f'Coulomb-counted SoC: {TEST_BATTERIES[0]} cycle 1')
ax.set_xlabel('Time (s)')
ax.set_ylabel('SoC (%)')
plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# 5. Train/validation/test split and scaling without leakage
# ============================================================================
train_df = raw_df[raw_df['battery_name'].isin(TRAIN_BATTERIES)].copy()
val_df = raw_df[raw_df['battery_name'].isin(VAL_BATTERIES)].copy()
test_df = raw_df[raw_df['battery_name'].isin(TEST_BATTERIES)].copy()

assert len(train_df) > 0 and len(val_df) > 0 and len(test_df) > 0, 'One split is empty.'

scaler_X = MinMaxScaler()
scaler_y = MinMaxScaler()

train_df[FEATURE_COLS] = scaler_X.fit_transform(train_df[FEATURE_COLS])
train_df['SOC_scaled'] = scaler_y.fit_transform(train_df[[TARGET_COL]])

val_df[FEATURE_COLS] = scaler_X.transform(val_df[FEATURE_COLS])
val_df['SOC_scaled'] = scaler_y.transform(val_df[[TARGET_COL]])

test_df[FEATURE_COLS] = scaler_X.transform(test_df[FEATURE_COLS])
test_df['SOC_scaled'] = scaler_y.transform(test_df[[TARGET_COL]])

print('Train rows:', len(train_df))
print('Validation rows:', len(val_df))
print('Test rows:', len(test_df))
```

```python
# ============================================================================
# 6. Build sequences inside each battery/cycle
# ============================================================================
def build_sequences(df, feature_cols, target_col='SOC_scaled', seq_length=30):
    X_all, y_all, meta_all = [], [], []

    for (battery, cycle), group in df.groupby(['battery_name', 'cycle']):
        group = group.sort_values('time')
        X_values = group[feature_cols].to_numpy(dtype=np.float32)
        y_values = group[target_col].to_numpy(dtype=np.float32)
        time_values = group['time'].to_numpy()

        if len(group) <= seq_length:
            continue

        for i in range(len(group) - seq_length):
            target_idx = i + seq_length
            X_all.append(X_values[i:target_idx])
            y_all.append(y_values[target_idx])
            meta_all.append((battery, int(cycle), float(time_values[target_idx])))

    return np.array(X_all, dtype=np.float32), np.array(y_all, dtype=np.float32), pd.DataFrame(meta_all, columns=['battery_name', 'cycle', 'time'])

X_train, y_train, train_meta = build_sequences(train_df, FEATURE_COLS, seq_length=SEQ_LENGTH)
X_val, y_val, val_meta = build_sequences(val_df, FEATURE_COLS, seq_length=SEQ_LENGTH)
X_test, y_test, test_meta = build_sequences(test_df, FEATURE_COLS, seq_length=SEQ_LENGTH)

print('X_train:', X_train.shape, 'y_train:', y_train.shape)
print('X_val:  ', X_val.shape, 'y_val:  ', y_val.shape)
print('X_test: ', X_test.shape, 'y_test: ', y_test.shape)
```

```python
# ============================================================================
# 7. Encoder-only Transformer for SoC regression
# ============================================================================
def positional_encoding(seq_len, d_model):
    positions = np.arange(seq_len)[:, np.newaxis]
    dims = np.arange(d_model)[np.newaxis, :]
    angles = positions / np.power(10000.0, (2 * (dims // 2)) / d_model)
    angles[:, 0::2] = np.sin(angles[:, 0::2])
    angles[:, 1::2] = np.cos(angles[:, 1::2])
    return tf.cast(angles[np.newaxis, :, :], tf.float32)


def transformer_encoder_block(x, d_model, num_heads, ffn_dim, dropout_rate, idx):
    attn = MultiHeadAttention(
        num_heads=num_heads,
        key_dim=d_model // num_heads,
        dropout=dropout_rate,
        name=f'mha_{idx}'
    )(x, x)
    x = Add(name=f'attn_residual_{idx}')([x, attn])
    x = LayerNormalization(epsilon=1e-6, name=f'attn_norm_{idx}')(x)

    ffn = Dense(ffn_dim, activation='relu', name=f'ffn_1_{idx}')(x)
    ffn = Dropout(dropout_rate, name=f'ffn_dropout_{idx}')(ffn)
    ffn = Dense(d_model, name=f'ffn_2_{idx}')(ffn)
    x = Add(name=f'ffn_residual_{idx}')([x, ffn])
    x = LayerNormalization(epsilon=1e-6, name=f'ffn_norm_{idx}')(x)
    return x


def build_soc_transformer(
    seq_len,
    n_features,
    d_model=64,
    num_heads=4,
    ffn_dim=128,
    num_layers=2,
    dropout_rate=0.2,
    learning_rate=3e-4,
):
    inputs = Input(shape=(seq_len, n_features), name='input_sequence')
    x = Dense(d_model, name='input_projection')(inputs)
    x = x + positional_encoding(seq_len, d_model)
    x = Dropout(dropout_rate, name='positional_dropout')(x)

    for i in range(num_layers):
        x = transformer_encoder_block(x, d_model, num_heads, ffn_dim, dropout_rate, i)

    x = Lambda(lambda t: t[:, -1, :], name='last_token')(x)
    x = Dense(64, activation='relu', name='regression_dense')(x)
    x = Dropout(dropout_rate, name='regression_dropout')(x)
    outputs = Dense(1, name='soc_output')(x)

    model = Model(inputs=inputs, outputs=outputs, name='SoC_Transformer_Encoder')
    model.compile(optimizer=Adam(learning_rate=learning_rate), loss='mse', metrics=['mae'])
    return model

transformer_model = build_soc_transformer(
    seq_len=SEQ_LENGTH,
    n_features=len(FEATURE_COLS),
    d_model=64,
    num_heads=4,
    ffn_dim=128,
    num_layers=2,
    dropout_rate=0.2,
    learning_rate=3e-4,
)

transformer_model.summary()
```

```python
# ============================================================================
# 8. Train
# ============================================================================
checkpoint_path = 'best_soc_transformer.keras'

callbacks = [
    EarlyStopping(monitor='val_loss', patience=12, restore_best_weights=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6, verbose=1),
    ModelCheckpoint(checkpoint_path, monitor='val_loss', save_best_only=True, verbose=1),
]

history = transformer_model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=100,
    batch_size=64,
    callbacks=callbacks,
    verbose=1,
)
```

```python
# ============================================================================
# 9. Plot training curves
# ============================================================================
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(history.history['loss'], label='Train loss', color='steelblue')
ax.plot(history.history['val_loss'], label='Validation loss', color='darkorange')
ax.set_title('Transformer Training Curve')
ax.set_xlabel('Epoch')
ax.set_ylabel('MSE on scaled SoC')
ax.legend()
plt.tight_layout()
plt.show()

if 'mae' in history.history:
    fig, ax = plt.subplots(figsize=(7, 4))
    ax.plot(history.history['mae'], label='Train MAE', color='steelblue')
    ax.plot(history.history['val_mae'], label='Validation MAE', color='darkorange')
    ax.set_title('Transformer MAE Curve')
    ax.set_xlabel('Epoch')
    ax.set_ylabel('MAE on scaled SoC')
    ax.legend()
    plt.tight_layout()
    plt.show()
```

```python
# ============================================================================
# 10. Evaluation helpers
# ============================================================================
def inverse_soc(y_scaled):
    return scaler_y.inverse_transform(np.asarray(y_scaled).reshape(-1, 1)).ravel()


def evaluate_predictions(name, y_true_scaled, y_pred_scaled):
    y_true = inverse_soc(y_true_scaled)
    y_pred = inverse_soc(y_pred_scaled)

    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    max_error = np.max(np.abs(y_true - y_pred))

    return {
        'Model': name,
        'RMSE': rmse,
        'MAE': mae,
        'R2': r2,
        'Max Error': max_error,
        'y_true': y_true,
        'y_pred': y_pred,
    }

val_pred = transformer_model.predict(X_val, verbose=0).ravel()
test_pred = transformer_model.predict(X_test, verbose=0).ravel()

val_results = evaluate_predictions('Transformer validation', y_val, val_pred)
test_results = evaluate_predictions('Transformer test', y_test, test_pred)

metrics_df = pd.DataFrame([
    {k: v for k, v in val_results.items() if k not in ['y_true', 'y_pred']},
    {k: v for k, v in test_results.items() if k not in ['y_true', 'y_pred']},
]).set_index('Model')

metrics_df.round(4)
```

```python
# ============================================================================
# 11. Plot test predictions
# ============================================================================
N_PLOT = min(2500, len(test_results['y_true']))

fig, ax = plt.subplots(figsize=(10, 4.5))
ax.plot(test_results['y_true'][:N_PLOT], label='Ground truth', color='steelblue', linewidth=1.2)
ax.plot(test_results['y_pred'][:N_PLOT], label='Estimated', color='tomato', linewidth=1.0, linestyle='--')
ax.set_title(f'SoC Prediction on Held-Out Battery {TEST_BATTERIES[0]}')
ax.set_xlabel('Sequence index')
ax.set_ylabel('SoC (%)')
ax.set_ylim(-5, 105)
ax.legend()
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(5, 5))
ax.scatter(test_results['y_true'], test_results['y_pred'], s=8, alpha=0.35, color='teal')
ax.plot([0, 100], [0, 100], color='black', linewidth=1, linestyle='--')
ax.set_title('Predicted vs True SoC')
ax.set_xlabel('True SoC (%)')
ax.set_ylabel('Predicted SoC (%)')
ax.set_xlim(-5, 105)
ax.set_ylim(-5, 105)
plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# 12. Error by cycle: useful for diagnosing bad regions
# ============================================================================
test_diag = test_meta.copy()
test_diag['y_true'] = test_results['y_true']
test_diag['y_pred'] = test_results['y_pred']
test_diag['abs_error'] = np.abs(test_diag['y_true'] - test_diag['y_pred'])

cycle_errors = test_diag.groupby(['battery_name', 'cycle']).agg(
    samples=('abs_error', 'size'),
    mae=('abs_error', 'mean'),
    max_error=('abs_error', 'max'),
).reset_index()

cycle_errors.sort_values('mae', ascending=False).head(15)
```

## Additional Models for Comparison

The following cells train LSTM, Bi-LSTM, and SVR using the same train/validation/test split as the Transformer. Each neural model gets its own training curves and prediction plot, and all models are compared together at the end.

```python
# ============================================================================
# 13. LSTM and Bi-LSTM models
# ============================================================================
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Bidirectional
from sklearn.svm import SVR


def build_lstm_model(seq_len, n_features, learning_rate=1e-4, dropout_rate=0.3):
    # Smaller and more regularized than the first LSTM version.
    # This is intended to reduce battery-specific overfitting.
    l2 = tf.keras.regularizers.l2(1e-4)

    model = Sequential(name='LSTM_SoC_Regularized')
    model.add(Input(shape=(seq_len, n_features)))
    model.add(LSTM(
        48,
        return_sequences=True,
        kernel_regularizer=l2,
        recurrent_regularizer=l2,
    ))
    model.add(Dropout(dropout_rate))
    model.add(LSTM(
        24,
        kernel_regularizer=l2,
        recurrent_regularizer=l2,
    ))
    model.add(Dropout(dropout_rate))
    model.add(Dense(16, activation='relu', kernel_regularizer=l2))
    model.add(Dropout(0.15))
    model.add(Dense(1))
    model.compile(optimizer=Adam(learning_rate=learning_rate), loss='mse', metrics=['mae'])
    return model


def build_bilstm_model(seq_len, n_features, learning_rate=1e-4, dropout_rate=0.35):
    # Regularized Bi-LSTM to reduce battery-specific overfitting.
    l2 = tf.keras.regularizers.l2(1e-4)

    model = Sequential(name='BiLSTM_SoC_Regularized')
    model.add(Input(shape=(seq_len, n_features)))
    model.add(Bidirectional(LSTM(
        40,
        return_sequences=True,
        kernel_regularizer=l2,
        recurrent_regularizer=l2,
    )))
    model.add(Dropout(dropout_rate))
    model.add(Bidirectional(LSTM(
        20,
        kernel_regularizer=l2,
        recurrent_regularizer=l2,
    )))
    model.add(Dropout(dropout_rate))
    model.add(Dense(16, activation='relu', kernel_regularizer=l2))
    model.add(Dropout(0.15))
    model.add(Dense(1))
    model.compile(optimizer=Adam(learning_rate=learning_rate), loss='mse', metrics=['mae'])
    return model

lstm_model = build_lstm_model(SEQ_LENGTH, len(FEATURE_COLS))
bilstm_model = build_bilstm_model(SEQ_LENGTH, len(FEATURE_COLS))

lstm_model.summary()
bilstm_model.summary()
```

```python
# ============================================================================
# 14. Train LSTM
# ============================================================================
lstm_callbacks = [
    EarlyStopping(monitor='val_loss', patience=8, restore_best_weights=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=4, min_lr=1e-6, verbose=1),
    ModelCheckpoint('best_soc_lstm.keras', monitor='val_loss', save_best_only=True, verbose=1),
]

history_lstm = lstm_model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=100,
    batch_size=64,
    callbacks=lstm_callbacks,
    verbose=1,
)
```

```python
# ============================================================================
# 18. Train improved SVR baseline
# ============================================================================
# SVR does not have layers like neural networks. We improve it by tuning:
# C, epsilon, gamma, and the number of training samples.

SVR_MAX_TRAIN = 80000
rng = np.random.default_rng(SEED)

X_train_flat = X_train.reshape(X_train.shape[0], -1)
X_val_flat = X_val.reshape(X_val.shape[0], -1)
X_test_flat = X_test.reshape(X_test.shape[0], -1)

if len(X_train_flat) > SVR_MAX_TRAIN:
    svr_idx = rng.choice(len(X_train_flat), size=SVR_MAX_TRAIN, replace=False)
    X_svr_train = X_train_flat[svr_idx]
    y_svr_train = y_train[svr_idx]
else:
    X_svr_train = X_train_flat
    y_svr_train = y_train

svr_model = SVR(
    kernel='rbf',
    C=100,
    epsilon=0.005,
    gamma='scale',
    cache_size=2000
)

svr_model.fit(X_svr_train, y_svr_train)

print('SVR training samples:', len(X_svr_train))
print('Improved SVR training complete.')
```

```python
# ============================================================================
# Inference time comparison
# ============================================================================
import time

def measure_inference_time(model_name, predict_fn, X_sample, n_runs=100):
    # Warm-up
    for _ in range(5):
        _ = predict_fn(X_sample)

    start = time.perf_counter()

    for _ in range(n_runs):
        _ = predict_fn(X_sample)

    end = time.perf_counter()

    avg_time_ms = ((end - start) / n_runs) * 1000

    return {
        'Model': model_name,
        'Average inference time (ms)': avg_time_ms
    }


# One sample for real-time deployment style
X_seq_sample = X_test[:1]       # shape: (1, 30, 4)
X_flat_sample = X_test_flat[:1] # shape: (1, 120)

inference_results = []

inference_results.append(
    measure_inference_time(
        'Transformer',
        lambda x: transformer_model.predict(x, verbose=0),
        X_seq_sample
    )
)

inference_results.append(
    measure_inference_time(
        'LSTM',
        lambda x: lstm_model.predict(x, verbose=0),
        X_seq_sample
    )
)

inference_results.append(
    measure_inference_time(
        'Bi-LSTM',
        lambda x: bilstm_model.predict(x, verbose=0),
        X_seq_sample
    )
)

inference_results.append(
    measure_inference_time(
        'SVR',
        lambda x: svr_model.predict(x),
        X_flat_sample
    )
)

inference_results.append(
    measure_inference_time(
        'Random Forest',
        lambda x: rf_model.predict(x),
        X_flat_sample
    )
)

inference_time_df = pd.DataFrame(inference_results).set_index('Model')
display(inference_time_df.round(4))


# Plot
fig, ax = plt.subplots(figsize=(8, 4))

sns.barplot(
    data=inference_time_df.reset_index(),
    x='Model',
    y='Average inference time (ms)',
    ax=ax,
    color='steelblue'
)

ax.set_title('Average Single-Sample Inference Time')
ax.set_xlabel('')
ax.set_ylabel('Time (ms)')
ax.tick_params(axis='x', rotation=20)

plt.tight_layout()
plt.savefig('soc_model_inference_time_comparison.png', dpi=200, bbox_inches='tight')
plt.show()

print('Saved: soc_model_inference_time_comparison.png')
```
